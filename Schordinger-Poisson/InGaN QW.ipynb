# Biblioteca de Materiales

```python
c_GaN = {
        'mass_e': 0.15,
        'mass_hh': 0.8,
        'Bandgap': 3.25
    }

c_InN = {
        'mass_e': 0.041,
        'mass_hh': 0.8,
        'Bandgap': 0.6
    }

def c_InGaN(x_In):
    return {
        'mass_e': x_In*c_InN['mass_e'] + (1-x_In)*c_GaN['mass_e'],
        'mass_hh': x_In*c_InN['mass_hh'] + (1-x_In)*c_GaN['mass_hh'],
        'Bandgap': x_In*c_InN['Bandgap'] + (1-x_In)*c_GaN['Bandgap'],
    }
```

# Set up

```python
# Espesor
l = 120

bandoffset = 0.7

barrier = c_GaN
well = c_InGaN(0.26)
```

# Calculo

```python
# Energía de la banda de conduccion

BC = bandoffset*(  barrier['Bandgap'] - well['Bandgap']  ) 
BV = (1-bandoffset)*(  barrier['Bandgap'] - well['Bandgap']  )

ec0 = efm.solucion(BC, barrier['mass_e'], well['mass_e'], l)
print(f"{l:.0f} Å \t {ec0*1000:.5f} meV")

# Energía de la banda de valencia
ev0 = efm.solucion(BV, barrier['mass_hh'], well['mass_hh'], l)
print(f"{l:.0f} Å \t {ev0*1000:.5f} meV")


ground_state_transition = ec0 + ev0 + well['Bandgap']
ground_state_transition
```
