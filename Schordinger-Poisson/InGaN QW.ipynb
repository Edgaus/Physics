# Biblioteca de Materiales

```python
c_GaN = {
        'mass_e': 0.15,
        'mass_hh': 0.8,
        'Bandgap': 3.225
    }

c_InN = {
        'mass_e': 0.041,
        'mass_hh': 0.8,
        'Bandgap': 0.6
    }

def c_InGaN(x_In):
    return {
        'mass_e': x_In*c_InN['mass_e'] + (1-x_In)*c_GaN['mass_e'],
        'mass_hh': x_In*c_InN['mass_hh'] + (1-x_In)*c_GaN['mass_hh'],
        'Bandgap': x_In*c_InN['Bandgap'] + (1-x_In)*c_GaN['Bandgap']-x_In*(1-x_In)*1.4,
    }
```

# Set up

```python
# Espesor
l = 120

bandoffset = 0.7


m914_In = [ 12.61, 8.62, 8.58 ]
m925_In = [ 10.9, 11.85, 12.43 ]
m930_In = [ 25.38, 18.3, 15.6 ]
m931_In = [ 26.3, 17.34, 15.14 ]


barrier = c_GaN
i=0
for x_In in m931_In:
    i +=1
    well = c_InGaN(x_In/100)
    BC = bandoffset*(  barrier['Bandgap'] - well['Bandgap']  ) 
    BV = (1-bandoffset)*(  barrier['Bandgap'] - well['Bandgap']  )

    print( barrier['Bandgap'], well['Bandgap'])

    ec0 = efm.solucion(BC, barrier['mass_e'], well['mass_e'], l)
    #print(f"{l:.0f} Å \t {ec0*1000:.5f} meV")

    # Energía de la banda de valencia
    ev0 = efm.solucion(BV, barrier['mass_hh'], well['mass_hh'], l)
    #print(f"{l:.0f} Å \t {ev0*1000:.5f} meV")


    ground_state_transition = ec0 + ev0 + well['Bandgap']
    print(f'La transicion del pozo:{i} es : {ground_state_transition} eV')


```
