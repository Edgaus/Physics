# Biblioteca de Materiales

```python
def AlGaAs(x_Al):
    return {
        'mass_e': 0.067 + 0.083*x_Al,
        'mass_hh': 0.62 + 0.14*x_Al,
        'Bandgap': 1.426 + 1.247*x_Al,
    }



GaAs = {
        'mass_e': 0.067,
        'mass_hh': 0.62,
        'Bandgap': 1.426
        }

c_GaN = {
        'mass_e': ,
        'mass_hh': 0.62,
        'Bandgap': 1.426
    }

c_InN = {
        'mass_e': 0.067,
        'mass_hh': 0.62,
        'Bandgap': 1.426
    }

def c_InGaN(x_In):
    return {
        'mass_e': x_In*c_InN['mass_e'] + (1-x_In)*c_GaN['mass_e'],
        'mass_hh': x_In*c_InN['mass_hh'] + (1-x_In)*c_GaN['mass_hh'],
        'Bandgap': x_In*c_InN['Bandgap'] + (1-x_In)*c_GaN['Bandgap'],
    }
```

# Set up

```python
# Espesor
l = 50

bandoffset = 0.7

barrier = AlGaAs(0.27)
well = GaAs

Temp =  65
```

# Temperature dependence Gap AlGaAs

```python
def Varshni_AlGaAs(x,T):
    A = 341.87+ 1.5194*T-(5.5E-4)*T**2
    B = 466.71 + 1.48*T -(3.3E-4)*T**2
    C = 130.24
    D = 225 + T
    E = 88

    return ( A + B*x + C*x**2   )/( D + E*x )
```

```python
# Energía de la banda de conduccion


BC = bandoffset*(   Varshni_AlGaAs(0.3,Temp)  - Varshni_AlGaAs(0,Temp)    ) 
BV = (1-bandoffset)*(   Varshni_AlGaAs(0.3,Temp)  -Varshni_AlGaAs(0,Temp)   )

# Energía de la banda de valencia
ev0 = efm.solucion(BV, barrier['mass_hh'], well['mass_hh'], l)
print(f"{l:.0f} Å \t {ev0*1000:.5f} meV")


ec0 = efm.solucion(BC, barrier['mass_e'], well['mass_e'], l)
print(f"{l:.0f} Å \t {ec0*1000:.5f} meV")



ground_state_transition = ec0 + ev0 + Varshni_AlGaAs(0,Temp) 
ground_state_transition
```

```python
# Espesor
l = 55

bandoffset = 0.67

barrier = AlGaAs(0.3)
well = GaAs
```

# Calculo

```python
# Energía de la banda de conduccion

BC = bandoffset*(  barrier['Bandgap'] - well['Bandgap']  ) 
BV = (1-bandoffset)*(  barrier['Bandgap'] - well['Bandgap']  )

ec0 = efm.solucion(BC, barrier['mass_e'], well['mass_e'], l)
print(f"{l:.0f} Å \t {ec0*1000:.5f} meV")

# Energía de la banda de valencia
ev0 = efm.solucion(BV, barrier['mass_hh'], well['mass_hh'], l)
print(f"{l:.0f} Å \t {ev0*1000:.5f} meV")


ground_state_transition = ec0 + ev0 + well['Bandgap']
ground_state_transition
```
